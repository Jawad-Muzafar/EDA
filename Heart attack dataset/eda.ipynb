# **Bi and Multivariate Eda**

## **Changing data type of result for heatmap**

```python
df['Result'].dtype
```

```python
# map textual labels to integers safely
orig = df['Result'].astype(str)
mapped = orig.str.strip().str.lower().map({'negative': 0, 'positive': 1})
if mapped.isnull().any():
    print("Unmapped Result values:", orig[mapped.isnull()].unique())

df['Result'] = mapped.astype('Int64')  # nullable integer dtype
```

```python
df['Result']
```

```python
sns.heatmap(df.select_dtypes('number').corr(), annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.show()
```
